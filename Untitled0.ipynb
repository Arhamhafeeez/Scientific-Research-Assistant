```python
# Final Project - Step 1
# Load a real open-science dataset

import pandas as pd

url = "https://raw.githubusercontent.com/plotly/datasets/master/earthquakes-23k.csv"

nasa_data = pd.read_csv(url)

print("REAL SCIENTIFIC DATA LOADED")
print("Rows:", len(nasa_data))
print("Columns:", len(nasa_data.columns))

print("\nFirst 5 records:")
print(nasa_data.head())
```

```python
# Final Project - Step 2
# Inspect the scientific dataset

print("DATASET INFORMATION")
nasa_data.info()

print("\nMISSING VALUES")
print(nasa_data.isnull().sum())

print("\nBASIC STATISTICS")
print(nasa_data.describe())
```

```python
# Final Project - Step 3
# Analyze earthquake magnitude

print("EARTHQUAKE ANALYSIS")

print("\nAverage Magnitude:")
print(nasa_data["Magnitude"].mean())

print("\nMinimum Magnitude:")
print(nasa_data["Magnitude"].min())

print("\nMaximum Magnitude:")
print(nasa_data["Magnitude"].max())

print("\nStrongest Earthquake Record:")
print(nasa_data.loc[nasa_data["Magnitude"].idxmax()])
```

```python
# Final Project - Step 4
# Create a Scientific Research Assistant

def scientific_assistant(question):
    question = question.lower()

    if "average magnitude" in question or "mean magnitude" in question:
        return f"The average earthquake magnitude is {nasa_data['Magnitude'].mean():.2f}."

    elif "minimum magnitude" in question:
        return f"The minimum earthquake magnitude is {nasa_data['Magnitude'].min():.1f}."

    elif "maximum magnitude" in question or "strongest" in question:
        return f"The maximum earthquake magnitude is {nasa_data['Magnitude'].max():.1f}."

    elif "records" in question or "rows" in question:
        return f"The dataset contains {len(nasa_data)} earthquake records."

    elif "columns" in question:
        return f"The dataset contains {len(nasa_data.columns)} columns: {', '.join(nasa_data.columns)}."

    else:
        return "I can answer questions about earthquake records, columns, average magnitude, minimum magnitude, and maximum magnitude."

print("SCIENTIFIC RESEARCH ASSISTANT READY!")
```

```python
# Final Project - Step 5
# Test the Scientific Research Assistant

question = input("Ask a question about the earthquake dataset: ")

answer = scientific_assistant(question)

print("\nANSWER:")
print(answer)
```

```python
# Final Project - Step 6
# Test different scientific questions

questions = [
    "How many records are in the dataset?",
    "What is the minimum magnitude?",
    "What is the maximum magnitude?",
    "How many columns are there?"
]

for question in questions:
    print("\nQUESTION:", question)
    print("ANSWER:", scientific_assistant(question))
```

```python
# Final Project - Step 7
# Add location information to the Scientific Research Assistant

def scientific_assistant_v2(question):
    question = question.lower()

    if "latitude" in question:
        return f"The dataset contains latitude information. Range: {nasa_data['Latitude'].min()} to {nasa_data['Latitude'].max()}."

    elif "longitude" in question:
        return f"The dataset contains longitude information. Range: {nasa_data['Longitude'].min()} to {nasa_data['Longitude'].max()}."

    elif "average magnitude" in question or "mean magnitude" in question:
        return f"The average earthquake magnitude is {nasa_data['Magnitude'].mean():.2f}."

    elif "minimum magnitude" in question:
        return f"The minimum earthquake magnitude is {nasa_data['Magnitude'].min():.1f}."

    elif "maximum magnitude" in question or "strongest" in question:
        return f"The maximum earthquake magnitude is {nasa_data['Magnitude'].max():.1f}."

    elif "records" in question or "rows" in question:
        return f"The dataset contains {len(nasa_data)} earthquake records."

    elif "columns" in question:
        return f"The dataset contains {len(nasa_data.columns)} columns: {', '.join(nasa_data.columns)}."

    else:
        return "I can answer questions about earthquake records, columns, latitude, longitude, and magnitude."

print("SCIENTIFIC RESEARCH ASSISTANT V2 READY!")
```

```python
# Final Project - Step 8
# Test location questions

questions = [
    "What is the latitude range?",
    "What is the longitude range?",
    "What is the maximum magnitude?"
]

for question in questions:
    print("\nQUESTION:", question)
    print("ANSWER:", scientific_assistant_v2(question))
```

```python
# Final Project - Step 9
# Add date information to the Scientific Research Assistant

def scientific_assistant_v3(question):
    question = question.lower()

    if "date" in question:
        return f"The dataset contains earthquake dates from {nasa_data['Date'].min()} to {nasa_data['Date'].max()}."

    elif "latitude" in question:
        return f"The latitude range is {nasa_data['Latitude'].min()} to {nasa_data['Latitude'].max()}."

    elif "longitude" in question:
        return f"The longitude range is {nasa_data['Longitude'].min()} to {nasa_data['Longitude'].max()}."

    elif "average magnitude" in question or "mean magnitude" in question:
        return f"The average earthquake magnitude is {nasa_data['Magnitude'].mean():.2f}."

    elif "minimum magnitude" in question:
        return f"The minimum earthquake magnitude is {nasa_data['Magnitude'].min():.1f}."

    elif "maximum magnitude" in question or "strongest" in question:
        return f"The maximum earthquake magnitude is {nasa_data['Magnitude'].max():.1f}."

    elif "records" in question or "rows" in question:
        return f"The dataset contains {len(nasa_data)} earthquake records."

    elif "columns" in question:
        return f"The dataset contains {len(nasa_data.columns)} columns: {', '.join(nasa_data.columns)}."

    else:
        return "I can answer questions about records, columns, dates, latitude, longitude, and earthquake magnitude."

print("SCIENTIFIC RESEARCH ASSISTANT V3 READY!")
```

```python
# Final Project - Step 10
# Final test of the Scientific Research Assistant

questions = [
    "What are the earthquake dates?",
    "What is the latitude range?",
    "What is the longitude range?",
    "What is the average magnitude?",
    "What is the strongest earthquake?"
]

for question in questions:
    print("\nQUESTION:", question)
    print("ANSWER:", scientific_assistant_v3(question))
```
